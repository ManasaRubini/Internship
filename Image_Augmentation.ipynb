```python
#slicing image
img2 = img[400:750,400:800]
img2
```

```python
#rotating images

r1_img = cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(r1_img)
```

```python
r2_img = cv2.rotate(img,cv2.ROTATE_180)
cv2_imshow(r2_img)
```

```python
r3_img = cv2.rotate(img,cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(r3_img)
```

```python
f1_img = cv2.flip(img,1)
cv2_imshow(f1_img)
```

```python
f2_img = cv2.flip(img,0)
cv2_imshow(f2_img)
```

```python
#resizing
r_img = cv2.resize(img,(600,400))
cv2_imshow(r_img)
```

```python
b_img = cv2.blur(img,(50,50))
cv2_imshow(b_img)
```
