```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load dataset
df = pd.read_csv('Mall_Customers.csv')

# Taking first 10 rows for clear visualization
df = df.head(10)

# X-axis positions
x = np.arange(len(df['CustomerID']))

# Width of each bar
width = 0.25

# Create grouped bars
plt.bar(x - width, df['Age'], width, label='Age')

plt.bar(x,
        df['Annual Income (k$)'],
        width,
        label='Annual Income')

plt.bar(x + width,
        df['Spending Score (1-100)'],
        width,
        label='Spending Score')

# Labels
plt.xlabel("Customer ID")
plt.ylabel("Values")
plt.title("Mall Customer Details")

# Customer IDs on x-axis
plt.xticks(x, df['CustomerID'])

# Legend
plt.legend()

# Show graph
plt.show()
```

```python
x
```

```python
# Create line chart
plt.plot(x,
         df['Age'],
         marker='o',
         label='Age')

plt.plot(x,
         df['Annual Income (k$)'],
         marker='o',
         label='Annual Income')

plt.plot(x,
         df['Spending Score (1-100)'],
         marker='o',
         label='Spending Score')

# Labels
plt.xlabel("Customer ID")
plt.ylabel("Values")
plt.title("Mall Customer Line Chart")

# Legend
plt.legend()

# Show graph
plt.show()
```

```python
values = [
    df['Age'].mean(),
    df['Annual Income (k$)'].mean(),
    df['Spending Score (1-100)'].mean(),
    df['Gender'].value_counts()['Male'],
    df['Gender'].value_counts()['Female']
]

labels = [
    'Average Age',
    'Average Income',
    'Average Spending Score',
    'Male Count',
    'Female Count'
]

# Pie chart
plt.pie(values,
        labels=labels,
        autopct='%1.1f%%')
# Title
plt.title("Average Customer Details")

# Show chart
plt.show()
```

```python
plt.scatter(df['Annual Income (k$)'],
            df['Spending Score (1-100)'])

# Labels
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")

# Title
plt.title("Income vs Spending Score")

# Show graph
plt.show()
```

```python
colors = df['Gender'].map({
    'Male': 'blue',
    'Female': 'red'
})
plt.scatter(
    df['Annual Income (k$)'],          # X-axis
    df['Spending Score (1-100)'],      # Y-axis
    s=df['Age'] * 5 , # Size based on Age
    c = colors
    )

# Labels
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")

# Title
plt.title("Income vs Spending Score")

# Show graph
plt.show()
```
