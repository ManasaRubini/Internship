```python
from apyori import apriori
```

```python
rules = apriori(transactions = transactions,min_support=0.003,min_confidence = 0.4,min_lift = 2,min_length = 2,max_length = 4 )
```

```python
result = list(rules)
```

```python
result
```

```python
def inspect(results):
    rows = []
    for result in results:
        for stat in result[2]:
            lhs        = ', '.join(sorted(stat.items_base)) if stat.items_base else 'N/A'
            rhs        = ', '.join(sorted(stat.items_add))  if stat.items_add  else 'N/A'
            support    = result.support
            confidence = stat.confidence
            lift       = stat.lift
            rows.append((lhs, rhs, support, confidence, lift))
    return rows

output_df = pd.DataFrame(
    inspect(result),
    columns=['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift']
)

output_df
```

```python

from pyECLAT import ECLAT
import pandas as pd

df_eclat = pd.DataFrame(transactions)

eclat_instance = ECLAT(data=df_eclat, verbose=True)

rule_indices, rule_supports = eclat_instance.fit(
    min_support   = 0.003,
    min_combination = 2,     # min_length
    max_combination = 2,     # max_length
    separator     = ' & ',
    verbose       = True
)

eclat_df = pd.DataFrame({
    'Rule'    : list(rule_supports.keys()),
    'Support' : list(rule_supports.values())
})

eclat_df.sort_values(by='Support', ascending=False).reset_index(drop=True)
```

```python
from mlxtend.frequent_patterns import fpgrowth, association_rules
from mlxtend.preprocessing   import TransactionEncoder
import pandas as pd

te      = TransactionEncoder()
te_arr  = te.fit(transactions).transform(transactions)
df_fp   = pd.DataFrame(te_arr, columns=te.columns_)

df_fp.head()
frequent_itemsets = fpgrowth(
    df_fp,
    min_support     = 0.003,
    use_colnames    = True,
    max_len         = 3       # max_length
)

frequent_itemsets
rules = association_rules(
    frequent_itemsets,
    metric          = 'lift',
    min_threshold   = 2       # min_lift
)

rules = rules[
    (rules['confidence'] >= 0.4) &   # min_confidence
    (rules['lift']       >= 2)
]

rules['antecedents'] = rules['antecedents'].apply(lambda x: ', '.join(sorted(x)))
rules['consequents'] = rules['consequents'].apply(lambda x: ', '.join(sorted(x)))

output_DataFrame = rules[[
    'antecedents',
    'consequents',
    'support',
    'confidence',
    'lift'
]].rename(columns={
    'antecedents' : 'Left Hand Side',
    'consequents' : 'Right Hand Side',
    'support'     : 'Support',
    'confidence'  : 'Confidence',
    'lift'        : 'Lift'
}).reset_index(drop=True)

output_DataFrame.sort_values(by='Lift', ascending=False).reset_index(drop=True)
```
