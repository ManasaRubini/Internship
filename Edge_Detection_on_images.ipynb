```python
import cv2
```

```python
img = cv2.imread('panda.jpg')
```

```python
gray_img = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray_img,110,110)
from google.colab.patches import cv2_imshow
cv2_imshow(edges)
```
